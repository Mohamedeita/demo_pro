```python
not_features = ["student_id","risk_level_rule" ,"next_term_gpa","next_cum_gpa","next_warning"]
```

```python
Featuers=[]
for i in df.columns:
    if i not in not_features:
        Featuers.append(i)
Featuers
```

```python
df_copy=df[Featuers].copy()
```

```python
df_copy.head()
```

```python
df_copy["prev_term_gpa"]=df_copy["prev_term_gpa"].fillna(df_copy["cum_gpa"])
df_copy["prev2_term_gpa"]=df_copy["prev2_term_gpa"].fillna(df_copy["prev_term_gpa"])
df_copy["gpa_trend"]=df_copy["gpa_trend"].fillna(0)
```

```python
X = df[Featuers].drop(
    columns=["risk_level_rule"],
    errors="ignore"
)

y_gpa = df["next_term_gpa"]
y_warning = df["next_warning"]
groups = df["student_id"]
```

```python
from sklearn.model_selection import GroupShuffleSplit

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y_warning, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_gpa_train = y_gpa.iloc[train_idx]
y_gpa_test = y_gpa.iloc[test_idx]

y_warning_train = y_warning.iloc[train_idx]
y_warning_test = y_warning.iloc[test_idx]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

preprocessor = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

gpa_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

gpa_model.fit(X_train_processed, y_gpa_train)

gpa_predictions = gpa_model.predict(X_test_processed)

print("Model trained successfully!")
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

y_gpa_pred = gpa_model.predict(X_test_processed)

r2 = r2_score(y_gpa_test, y_gpa_pred)
mae = mean_absolute_error(y_gpa_test, y_gpa_pred)
rmse = np.sqrt(mean_squared_error(y_gpa_test, y_gpa_pred))

print("R2 Score:", r2)
print("MAE:", mae)
print("RMSE:", rmse)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

gb_model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

gb_model.fit(X_train_processed, y_gpa_train)

gb_pred = gb_model.predict(X_test_processed)

print("R2:", r2_score(y_gpa_test, gb_pred))
print("MAE:", mean_absolute_error(y_gpa_test, gb_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_gpa_test, gb_pred)))
```
